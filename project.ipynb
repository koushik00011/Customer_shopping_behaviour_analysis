```python
import pandas as pd
df=pd.read_csv("C:\sql\customer_shopping_behavior.csv")
```

```python
df.head()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df.columns=df.columns.str.strip()
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.head()
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df=df.drop('promo_code_used',axis=1)
```

```python
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['review_rating']=df.groupby('category') ['review_rating'].transform(lambda x:x.fillna(x.median()))
```

```python

df.isnull().sum()
```

```python
df.head()
```

```python
label=['Young','Adult','Middle_agr','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=label)
```

```python
df.head()
```

```python
df['frequency_of_purchases'].unique()
```

```python
frequency={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90}
df['frequency_of_purchases_in_days']=df['frequency_of_purchases'].map(frequency)
```
